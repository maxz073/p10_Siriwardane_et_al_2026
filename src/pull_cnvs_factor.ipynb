# Pull FUT_CNVS_FACTOR for Treasury futures

Pulls only the conversion factor field for the futures contracts defined in `pull_bloomberg.py`. Requires Bloomberg terminal and `xbbg`.

```python
from pathlib import Path

import pandas as pd

# Reuse contracts and dates from pull_bloomberg (run with notebook in src/)
from pull_bloomberg import FUTURES_CONTRACTS, START_DATE, END_DATE
```

```python
from xbbg import blp

FIELD = "FUT_CNVS_FACTOR"

# bdp = point-in-time (one value per contract)
cnvs_bdp = blp.bdp(FUTURES_CONTRACTS, FIELD)
cnvs_bdp
```

Optional: get **historical** conversion factor via `bdh` (time series over the date range). Uncomment and run the cell below if you need a time series.

```python
# Historical FUT_CNVS_FACTOR (bdh)
cnvs_frames = []
for fut in FUTURES_CONTRACTS:
    df = blp.bdh(fut, [FIELD], START_DATE, END_DATE)
    cnvs_frames.append(df)
cnvs_bdh = pd.concat(cnvs_frames, axis=1)
cnvs_bdh.index.name = "Date"
cnvs_bdh
```
