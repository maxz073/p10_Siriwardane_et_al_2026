```python
# Implied repo from calc_spread.py (first deferred contract, volume > 0)
df_irr = pl.read_parquet("../data_manual/implied_repo_first_deferred.parquet")
```

```python
import numpy as np
import pandas as pd

tenors = ["2Y", "5Y", "10Y", "20Y", "30Y"]
pdf = df_irr.to_pandas()
dates = pd.to_datetime(pdf["Date"]).dt.normalize()
day = dates.dt.day
last_day = (dates + pd.offsets.MonthEnd(0)).dt.day
in_first_3 = (day <= 3).values
in_last_3 = (day >= last_day - 2).values
in_first_or_last_3 = in_first_3 | in_last_3

total_spikes = 0
spikes_in_window = 0
for col in tenors:
    s = pdf[col].dropna()
    if len(s) < 2:
        continue
    mean, std = s.mean(), s.std()
    if std == 0:
        continue
    is_spike = (~pdf[col].isna()) & (np.abs(pdf[col] - mean) > 3 * std)
    total_spikes += is_spike.sum()
    spikes_in_window += (is_spike & in_first_or_last_3).sum()

pct = 100 * spikes_in_window / total_spikes if total_spikes > 0 else float("nan")
print(f"Percentage of big spikes (>3 sd) in first or last 3 days of month: {pct:.2f}%")
print(f"  (spikes in window: {spikes_in_window}, total spikes: {total_spikes})")
```

```python
# List the top 10 biggest spikes in absolute value across all tenors (excluding NaN)
top_spikes = []
for col in tenors:
    s = pdf[["Date", col]].dropna()
    if len(s) == 0:
        continue
    mean = s[col].mean()
    std = s[col].std()
    s["abs_dev"] = (s[col] - mean).abs()
    s["tenor"] = col
    top_spikes.append(s)

if top_spikes:
    all_spikes = pd.concat(top_spikes, ignore_index=True)
    top10 = all_spikes.sort_values("abs_dev", ascending=False).head(20)
    print("Top 10 biggest spikes (by absolute deviation from mean, across all tenors):")
    display(top10[["Date", "tenor", col, "abs_dev"]])
else:
    print("No valid spikes found.")
```

```python
import matplotlib.pyplot as plt

# Define the date range
start_date = '2010-01-01'
end_date = '2024-02-28'

# Ensure 'date' is in datetime format
df_arb_spread['date'] = pd.to_datetime(df_arb_spread['date'])

# Filter for selected date range
mask = (df_arb_spread['date'] >= start_date) & (df_arb_spread['date'] <= end_date)
df_filtered = df_arb_spread.loc[mask].copy()
date_range = df_filtered['date']

# Properly align "mine" and "target" to the filtered DataFrame by date
# Assume mine/target are Series or arrays covering the same full period as df_arb_spread
# If they are not indexed by date, extract the appropriate values from original DataFrame within date_range
# The code will try to handle both "indexed by date" or just arrays

def extract_aligned(series_like, filtered_df, colname=None):
    import pandas as pd
    import numpy as np
    # If it's a Series with matching index or column name, align on date
    if isinstance(series_like, pd.Series):
        # If the index is date or same shape, try to align by index
        try:
            aligned = series_like.loc[filtered_df.index]
            return aligned.values
        except Exception:
            pass
    # If given as a DataFrame and colname is specified
    if isinstance(series_like, pd.DataFrame) and colname:
        try:
            return series_like.loc[filtered_df.index, colname].values
        except Exception:
            pass
    # If mine/target are simply arrays, slice according to length (WARNING: this assumes same start alignment!)
    arr = getattr(series_like, "values", None)
    if arr is None:
        arr = getattr(series_like, "to_numpy", lambda: np.asarray(series_like))()
    # If filtering by date reduced DataFrame, match up by slicing or pad if needed
    if len(arr) >= len(filtered_df):
        return arr[:len(filtered_df)]
    else:
        return np.pad(arr, (0, len(filtered_df) - len(arr)), constant_values=np.nan)

# If mine/target are actually columns in df_arb_spread, pull from filtered version
mine_in_df = 'mine' in df_filtered.columns
target_in_df = 'target' in df_filtered.columns
if mine_in_df:
    mine_aligned = df_filtered['mine'].values
else:
    mine_aligned = extract_aligned(mine, df_filtered)
if target_in_df:
    target_aligned = df_filtered['target'].values
elif isinstance(target, pd.Series) and (target.index.equals(df_arb_spread.index) or target.index.equals(df_arb_spread['date'])):
    # Likely same length, just slice using the mask
    target_aligned = target[mask].values
else:
    # Try just aligning by position
    target_aligned = extract_aligned(target, df_filtered)

# Make sure all columns have the same length
min_len = min(len(date_range), len(mine_aligned), len(target_aligned))

plot_df = pd.DataFrame({
    "date": date_range.iloc[:min_len].values,
    "mine": mine_aligned[:min_len],
    "target": target_aligned[:min_len]
})

# Drop NA rows for plotting purposes (optional: drop where both mine and target are nan)
plot_df = plot_df.dropna(subset=["mine", "target"], how="all")

# -- Debug: Check for completely empty or all-NA target data
if plot_df['target'].isna().all():
    print("WARNING: All values in 'target' are NaN after alignment and filtering. Check target length/index or how you extract target values.")
    print(f"target_aligned shape: {target_aligned.shape}")
    # Optionally print some sample of target_aligned for debugging
    print("First 10 of target_aligned:", target_aligned[:10])

plt.figure(figsize=(12, 6))
plt.plot(plot_df['date'], plot_df['mine'], label='mine')
plt.plot(plot_df['date'], plot_df['target'], label='target')
plt.xlabel('Date')
plt.ylabel('Value')
plt.title('Mine & Target over Time (2010-01-01 to 2024-02-28)')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

# Compute RMSE between 'mine' and 'target' over 2010-01-01 to 2024-02-28
mine_vals = plot_df['mine'].values
target_vals = plot_df['target'].values

# Drop pairs where either value is nan
mask = ~((np.isnan(mine_vals)) | (np.isnan(target_vals)))
if mask.sum() == 0:
    print("No valid data to compute RMSE (all values are NaN or masked).")
else:
    rmse = np.sqrt(mean_squared_error(mine_vals[mask], target_vals[mask]))
    print(f"RMSE between 'mine' and 'target' (2010-01-01 to 2024-02-28): {rmse:.4f}")
```

```python
# Calculate correlation between 'mine' and 'target' in plot_df
# Drop pairs where either is nan
mine_vals = plot_df['mine'].values
target_vals = plot_df['target'].values
mask = ~((np.isnan(mine_vals)) | (np.isnan(target_vals)))

if mask.sum() == 0:
    print("No valid data to compute correlation (all values are NaN or masked).")
else:
    corr = np.corrcoef(mine_vals[mask], target_vals[mask])[0, 1]
    print(f"Correlation between 'mine' and 'target' (2010-01-01 to 2024-02-28): {corr:.4f}")
```

```python
# Compare my IRR to Bloomberg's IRR for all contracts (2010–2020)
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

# Date range (up to 2020)
start_date = pl.date(2010, 1, 1)
end_date = pl.date(2020, 1, 1)
WINDOW_SIZE = 7

# Tenor -> Bloomberg ticker prefix (for finding fut_implied_repo_rt column)
TENOR_BBG = {"2Y": "TU2", "5Y": "FV2", "10Y": "TY2", "20Y": "WN2", "30Y": "US2"}
tenors = list(TENOR_BBG.keys())

# My IRR: filter once, cast Date, only up to 2020
my_irr_all = (
    df_irr.filter(pl.col("Date").is_between(start_date, end_date, closed="left"))
    .with_columns(pl.col("Date").cast(pl.Date))
)

# Bloomberg: filter once by date, only up to 2020
df_bbg = df.filter(pl.col("Date").is_between(start_date, end_date, closed="left"))

fig, axes = plt.subplots(5, 1, figsize=(10, 12), sharex=True)
results = []

for i, tenor in enumerate(tenors):
    bbg_prefix = TENOR_BBG[tenor]
    bbg_col = [c for c in df.columns if bbg_prefix in str(c) and "fut_implied_repo_rt" in str(c)]
    if not bbg_col:
        axes[i].set_title(f"{tenor}: no Bloomberg column")
        continue
    bbg_col = bbg_col[0]

    my_irr = my_irr_all.select(["Date", pl.col(tenor).alias("my_irr_bps")])
    bbg_irr = df_bbg.select([
        "Date",
        (pl.col(bbg_col) * 100).alias("bloomberg_irr_bps"),  # percent -> bps
    ])
    compare = my_irr.join(bbg_irr, on="Date", how="inner")
    compare_clean = compare.filter(
        pl.col("my_irr_bps").is_not_nan() & pl.col("bloomberg_irr_bps").is_not_nan()
    )
    # 5-day rolling median for smoother display
    compare_clean = compare_clean.sort("Date").with_columns([
        pl.col("my_irr_bps").rolling_median(window_size=WINDOW_SIZE).alias("my_irr_bps_smooth"),
        pl.col("bloomberg_irr_bps").rolling_median(window_size=WINDOW_SIZE).alias("bloomberg_irr_bps_smooth"),
    ])

    ax = axes[i]
    ax.plot(compare_clean["Date"], compare_clean["my_irr_bps_smooth"], label=f"My IRR ({tenor})", alpha=0.8)
    ax.plot(compare_clean["Date"], compare_clean["bloomberg_irr_bps_smooth"], label=f"Bloomberg ({tenor})", alpha=0.8)
    ax.set_ylabel("bps")
    ax.set_title(f"{tenor}")
    ax.legend(loc="upper right", fontsize=8)
    ax.grid(True, alpha=0.3)

    my_vals = compare_clean["my_irr_bps"].to_numpy()
    bbg_vals = compare_clean["bloomberg_irr_bps"].to_numpy()
    n = len(my_vals)
    if n > 0:
        rmse = (mean_squared_error(my_vals, bbg_vals)) ** 0.5
        corr = np.corrcoef(my_vals, bbg_vals)[0, 1]
        results.append({"tenor": tenor, "n": n, "rmse_bps": rmse, "corr": corr})
    else:
        results.append({"tenor": tenor, "n": 0, "rmse_bps": np.nan, "corr": np.nan})

axes[-1].set_xlabel("Date")
fig.suptitle("My IRR vs Bloomberg IRR (2010–2020)", y=1.02)
plt.tight_layout()
plt.show()

# Summary table
print("2010–2026 comparison by contract:")
print(pl.DataFrame(results))
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
import pandas as pd

print("RMSE of normalized (z-scored) my_irr_bps vs bloomberg_irr_bps for each tenor (pre-2020 only):")
for i, tenor in enumerate(tenors):
    bbg_prefix = TENOR_BBG[tenor]
    bbg_col = [c for c in df.columns if bbg_prefix in str(c) and "fut_implied_repo_rt" in str(c)]
    if not bbg_col:
        print(f"{tenor}: no Bloomberg column")
        continue
    bbg_col = bbg_col[0]

    my_irr = my_irr_all.select(["Date", pl.col(tenor).alias("my_irr_bps")])
    bbg_irr = df_bbg.select([
        "Date",
        (pl.col(bbg_col) * 100).alias("bloomberg_irr_bps"),  # percent -> bps
    ])
    compare = my_irr.join(bbg_irr, on="Date", how="inner")
    compare_clean = compare.filter(
        pl.col("my_irr_bps").is_not_nan() & pl.col("bloomberg_irr_bps").is_not_nan()
    )

    # Filter for only pre-2020 dates
    compare_clean = compare_clean.filter(pl.col("Date") < pd.Timestamp("2020-01-01"))

    compare_clean_pd = compare_clean.to_pandas()

    if compare_clean_pd.shape[0] == 0:
        print(f"{tenor}: no data")
        continue

    # Normalize the two IRR columns separately (z-score normalization)
    scaler = StandardScaler()
    normalized = scaler.fit_transform(compare_clean_pd[["my_irr_bps", "bloomberg_irr_bps"]])

    my_irr_norm = normalized[:, 0]
    bbg_irr_norm = normalized[:, 1]

    # Compute RMSE of the normalized values
    rmse_norm = (mean_squared_error(my_irr_norm, bbg_irr_norm)) ** 0.5

    print(f"{tenor}: RMSE = {rmse_norm:.4f}")
```
