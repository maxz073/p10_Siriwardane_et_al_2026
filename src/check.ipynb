```python
import pandas as pd

# Try reading the file using pandas.read_stata (since the extension is .dta, likely Stata format)
df_sf_rf = pd.read_stata("../data_manual/treasury_sf_implied_rf.dta")
```

```python
# 1.1 — spread_n = tfut_n_rf - tfut_ois_n for each tenor
for n in [2, 5, 10, 20, 30]:
    df_sf_rf[f"spread_{n}"] = df_sf_rf[f"tfut_{n}_rf"] - df_sf_rf[f"tfut_ois_{n}"]
```

```python
df_sf_rf.tail(200)
```

```python
df_sf_rf_mean = df_sf_rf.mean()
df_sf_rf_median = df_sf_rf.median()
print("Column means:")
print(df_sf_rf_mean)
print("\nColumn medians:")
print(df_sf_rf_median)
```
